```python
import hashlib
def calc_sha256(input_string):
  return hashlib.sha256(input_string.encode('utf-8')).hexdigest()
string01 = "Cryptography is the mathematics of secrets"
# "." at the end of 2nd string
string02 = "Cryptography is the mathematics of secrets."

hash01 = calc_sha256(string01)
hash02 = calc_sha256(string02)
print(f"string01 : '{string01}'")
print(f"SHA-256 -01 :{hash01}")

print(f"\nstring02 :'{string02}'")
print(f"SHA 256-02 :{hash02 }")

# Task 2: compare the hashes
same_hex = sum(a == b for a, b in zip(hash01, hash02))

print("\nIdentical hashes?        ", hash01 == hash02)
print("Matching hex characters: ", same_hex, "/ 64")
print("Differing hex characters:", 64 - same_hex, "/ 64")

#Task3
def changed_bits(hash_a, hash_b):
    # XOR gives a 1 in every bit position where the two hashes differ
    diff = int(hash_a, 16) ^ int(hash_b, 16)
    return bin(diff).count('1')
def percent_changed_bits(hash_a, hash_b):
    total_bits = len(hash_a) * 4        # 64 hex characters x 4 bits = 256 bits
    return changed_bits(hash_a, hash_b) / total_bits * 100

print("\nChanged bits:", changed_bits(hash01, hash02), "/ 256")
print(f"Percentage of changed bits: {percent_changed_bits(hash01, hash02):.2f}%")
```
